```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#  Load House Dataset
house_df = pd.read_csv(r"C:\Users\kamal\OneDrive\Desktop\Work done\seaborn assignment\house.csv.csv")
house_df.head()
```

```python
# Task 8: Multi-Plots
g = sns.FacetGrid(house_df, col="Id")
g.map(sns.scatterplot, "Id", "SalePrice")
plt.show()

sns.relplot(x="Id", y="SalePrice", data=house_df)
plt.show()

sns.catplot(x="Id", y="SalePrice", kind="box", data=house_df)
plt.show()

sns.displot(house_df["SalePrice"], kde=True)
plt.show()
```
