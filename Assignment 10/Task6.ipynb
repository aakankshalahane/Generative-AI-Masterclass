###Task 6: Filtering & Conditional Selection
Filter students:

Marks > 75.

Subject = Math.

Marks > average.

Failed (marks < 70).

```python
import pandas as pd

students = {
    'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
    'Marks': [78, 85, 90, 66, 72],
    'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
}

df = pd.DataFrame(students)


print("Marks > 75:\n", df[df["Marks"] > 75])
print("Subject = Math:\n", df[df["Subject"] == "Math"])
print("Marks > Average:\n", df[df["Marks"] > df["Marks"].mean()])
print("Failed (Marks < 70):\n", df[df["Marks"] < 70])
```
