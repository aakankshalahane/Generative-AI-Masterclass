###Task 1: Pandas Series Basics
Import Pandas (import pandas as pd).

Create a Series from list: marks = [78, 85, 90, 66, 72].

Print values, index, and data type.

Access first element and last two elements.

```python
import pandas as pd
marks = [78, 85, 90, 66, 72]
series = pd.Series(marks)

print("Series values:\n", series.values)
print("Index:\n", series.index)
print("Data type:", series.dtype)

print("First element:", series.iloc[0])
print("Last two elements:\n", series.iloc[-2:])
```
