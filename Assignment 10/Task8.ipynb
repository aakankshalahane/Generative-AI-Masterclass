###Task 8: Pandas Plotting (Simple Graphs)
Plot using built-in Pandas plotting:

Bar graph: Names vs Marks.

Line graph of marks.

Histogram of marks.

Use DataFrame.plot() or Series.plot().

```python

import pandas as pd
import matplotlib.pyplot as plt

students = {
    'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
    'Marks': [78, 85, 90, 66, 72],
    'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
}

df = pd.DataFrame(students)

print("Average marks per subject:\n", df.groupby("Subject")["Marks"].mean())
print("Count per subject:\n", df.groupby("Subject")["Name"].count())
print("Max marks per subject:\n", df.groupby("Subject")["Marks"].max())

# Bar graph: Names vs Marks
df.plot(x="Name", y="Marks", kind="bar", title="Marks by Student")

# Line graph of marks
df["Marks"].plot(kind="line", title="Marks Line Graph")

# Histogram of marks
df["Marks"].plot(kind="hist", title="Marks Histogram")
```
