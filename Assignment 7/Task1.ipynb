###Task 1: Basic Class & Object Creation
Create a Product class with attributes: name, price, category.

Method: get_info() → prints product details.

Extra: apply_discount(percent) → returns discounted price.

Concepts: Classes, objects, methods.

```python
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.price = price
        self.category = category

    def get_info(self):
        print(f"Product: {self.name}, Price: {self.price}, Category: {self.category}")

    def apply_discount(self, percent):
        return self.price - (self.price * percent / 100)

# Test
p1 = Product("Laptop", 50000, "Electronics")
p2 = Product("Mobile", 20000, "Electronics")
p1.get_info()
p2.get_info()
print("Discounted Laptop Price:", p1.apply_discount(10))
```
