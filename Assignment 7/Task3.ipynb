###Task 3: Inheritance
Create subclass ElectronicProduct inheriting from Product.

Add attribute: warranty_years.

Override get_info() to include warranty details.

Concepts: Inheritance, method overriding.

```python
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.price = price
        self.category = category

    def get_info(self):
        print(f"Product: {self.name}, Price: {self.price}, Category: {self.category}")


class ElectronicProduct(Product):
    def __init__(self, name, price, category, warranty_years):
        super().__init__(name, price, category)
        self.warranty_years = warranty_years

    def get_info(self):
        print(f"{self.name} ({self.category}) - Price: {self.price}, Warranty: {self.warranty_years} years")


# Test
ep = ElectronicProduct("TV", 40000, "Electronics", 2)
ep.get_info()
```
