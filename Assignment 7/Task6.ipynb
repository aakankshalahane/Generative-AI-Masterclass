###Task 6: Magic Methods & Operator Overloading
Add __str__ method to Product → returns readable string.

Add __add__ method → allows product1 + product2 to return combined price.

Concepts: Magic methods, operator overloading

```python
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.price = price
        self.category = category

    # __str__ gives a nice readable output when you print the object
    def __str__(self):
        return f"Product({self.name}, {self.price}, {self.category})"

    # __add__ allows using + between two Product objects
    def __add__(self, other):
        return self.price + other.price


# ---------------- Testing ----------------
p1 = Product("Laptop", 50000, "Electronics")
p2 = Product("Mobile", 20000, "Electronics")

print(p1)  # Uses __str__
print(p2)  # Uses __str__

# Operator overloading
print("Combined Price:", p1 + p2)
```
