###Task 7: Mini Project – Inventory System
Create Inventory class to manage multiple products.

Methods: add_product, remove_product, total_value, show_products.

Create Store class with store name and inventory object.

Methods: add_new_product, show_summary.

Demonstrate by creating a store, adding products, showing summary, and testing operator overloading.

Concepts: Aggregation, system design, combining OOP principles.

```python
# Product class (with operator overloading from Task 6)
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.price = price
        self.category = category

    def get_price(self):
        return self.price

    def __str__(self):
        return f"Product({self.name}, {self.price}, {self.category})"

    def __add__(self, other):
        return self.price + other.price


# Inventory class
class Inventory:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def remove_product(self, product_name):
        self.products = [p for p in self.products if p.name != product_name]

    def total_value(self):
        return sum(p.get_price() for p in self.products)

    def show_products(self):
        for p in self.products:
            print(p)


# Store class
class Store:
    def __init__(self, name):
        self.name = name
        self.inventory = Inventory()

    def add_new_product(self, product):
        self.inventory.add_product(product)

    def show_summary(self):
        print(f"Store: {self.name}")
        self.inventory.show_products()
        print("Total Inventory Value:", self.inventory.total_value())

#Test
if __name__ == "__main__":
    # Create store
    store = Store("TechWorld")

    # Create products
    p1 = Product("Laptop", 50000, "Electronics")
    p2 = Product("Mobile", 20000, "Electronics")
    p3 = Product("Headphones", 3000, "Accessories")

    # Add products to store
    store.add_new_product(p1)
    store.add_new_product(p2)
    store.add_new_product(p3)

    # Show summary
    store.show_summary()

    # Test operator overloading
    print("Combined price of Laptop + Mobile:", p1 + p2)
```
