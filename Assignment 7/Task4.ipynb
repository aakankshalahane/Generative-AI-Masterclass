###Task 4: Polymorphism
Create subclasses Laptop and Mobile inheriting from Product.

Both override get_info() with their own style.

Demonstrate polymorphism by looping through objects and calling get_info().

Concepts: Polymorphism, dynamic method binding.

```python
# Base class
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.price = price
        self.category = category

    def get_info(self):
        print(f"Product: {self.name}, Price: {self.price}, Category: {self.category}")


# Subclasses overriding get_info
class Laptop(Product):
    def get_info(self):
        print(f"Laptop: {self.name}, Price: {self.price}")


class Mobile(Product):
    def get_info(self):
        print(f"Mobile: {self.name}, Price: {self.price}")


# Test polymorphism
items = [
    Laptop("Dell", 60000, "Electronics"),
    Mobile("iPhone", 80000, "Electronics")
]

for item in items:
    item.get_info()
```
