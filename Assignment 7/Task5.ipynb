###Task 5: Abstraction
Create abstract class Payment with abstract method process_payment(amount).

Subclasses: CreditCardPayment, UPIPayment.

Each implements process_payment() with simple print statements.

Concepts: Abstraction, abstract base classes

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def process_payment(self, amount):
        pass

class CreditCardPayment(Payment):
    def process_payment(self, amount):
        print(f"Paid {amount} using Credit Card.")

class UPIPayment(Payment):
    def process_payment(self, amount):
        print(f"Paid {amount} using UPI.")

# Test
cc = CreditCardPayment()
upi = UPIPayment()
cc.process_payment(5000)
upi.process_payment(2000)
```
