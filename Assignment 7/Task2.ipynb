###Task 2: Constructor & Encapsulation
Modify Product to make price a private attribute (__price).

Add getter (get_price) and setter (set_price) methods.

Setter ensures price is updated only if positive.

Concepts: Encapsulation, data hiding, safe attribute access.

```python
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.__price = price   # private attribute
        self.category = category

    def get_info(self):
        print(f"Product: {self.name}, Price: {self.__price}, Category: {self.category}")

    def get_price(self):
        return self.__price

    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price
        else:
            print("Invalid price!")

# Test
p = Product("Headphones", 3000, "Accessories")
p.get_info()
p.set_price(3500)
print("Updated Price:", p.get_price())
```
