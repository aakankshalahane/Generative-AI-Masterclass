```python
# Task 3: Product Pricing (Dictionaries)
```

```python
# 1. Create a dictionary with product names and prices
price_dict = {
    "laptop": 50000,
    "mobile": 25000,
    "keyboard": 2000,
    "mouse": 1000,
    "monitor": 12000,
    "headphone": 3000
}
```

```python
print("Original prices:", price_dict)
```

```python
# 2. Add a new product
price_dict["tablet"] = 18000
```

```python
print("After adding tablet:", price_dict)
```

```python
# Update the price of an existing product
price_dict["mouse"] = 1200
print("After updating mouse price:", price_dict)
```

```python
# Remove a product
product = "headphone"

if product in price_dict:
    del price_dict[product]
    print("After removing headphone:", price_dict)
else:
    print("Product does not exist")
```

```python
# 3. Calculate the average price
total = sum(price_dict.values())
number_of_products = len(price_dict)

average_price = total / number_of_products

print("Average price:", average_price)
```
