```python
# Categories matching each product
categories = ["Electronics", "Electronics", "Electronics", "Electronics", "Electronics", "Electronics"]
```

```python
categories
```

```python
# Create a set of categories
categories_set = set(categories)
```

```python
print("Categories:", categories_set)
```

```python
# Add a new category
categories_set.add("Accessories")
```

```python
print("After adding new category:", categories_set)
```

```python
# Add a duplicate category
categories_set.add("Electronics")
```

```python
print("After adding duplicate:", categories_set)
```

```python
# Check if a category exists
print("Is Electronics present?", "Electronics" in categories_set)
```

```python
# Total number of unique categories
print("Total unique categories:", len(categories_set))
```
