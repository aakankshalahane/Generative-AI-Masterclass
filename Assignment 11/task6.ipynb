Task 6: Histogram (Marks Distribution)
Dataset: Numeric dataset (e.g., student marks, ages).

Steps:

Create histogram showing distribution.

Choose appropriate number of bins.

Add title and axis labels.

```python
import matplotlib.pyplot as plt

marks = [55, 60, 65, 70, 75, 80, 85, 90, 95, 100]

plt.hist(marks, bins=5)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```
