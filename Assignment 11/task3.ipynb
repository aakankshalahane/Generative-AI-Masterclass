##Task 3: Bar Plot (Vertical & Horizontal)
Dataset: Categorical dataset (e.g., product categories with sales).

Steps:

Create a vertical bar chart.

Create a horizontal bar chart for the same data.

Add titles and labels.

```python
import matplotlib.pyplot as plt

categories = ["A", "B", "C", "D"]
values = [4, 7, 3, 8]

# Vertical
plt.bar(categories, values)
plt.title("Vertical Bar Plot")
plt.xlabel("Category")
plt.ylabel("Value")
plt.show()

# Horizontal
plt.barh(categories, values)
plt.title("Horizontal Bar Plot")
plt.xlabel("Value")
plt.ylabel("Category")
plt.show()
```
