##Task 5: Stacked Bar Chart
Dataset: Data with multiple categories contributing to totals (e.g., quarterly sales split by product).

Steps:

Create stacked bars (one category stacked on another).

Add legend and labels.

Display the plot.

```python
import matplotlib.pyplot as plt

categories = ["Q1", "Q2", "Q3", "Q4"]
sales_A = [30, 40, 50, 60]
sales_B = [20, 25, 30, 35]

plt.bar(categories, sales_A, label="Product A")
plt.bar(categories, sales_B, bottom=sales_A, label="Product B")

plt.title("Stacked Sales by Quarter")
plt.xlabel("Quarter")
plt.ylabel("Sales")
plt.legend()
plt.show()
```
