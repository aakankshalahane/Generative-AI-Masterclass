###Task 1: Line Plot (Sales Trend)
Dataset: Sales dataset (e.g., monthly sales from Kaggle or mock data).

Steps:

Plot sales values against months.

Add title, X-axis label (Months), Y-axis label (Sales).

Display the plot.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 170, 160, 180, 200]

plt.plot(months, sales, marker='o')
plt.title("Sales Trend")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```
