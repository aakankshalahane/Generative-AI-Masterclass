###Task 4: Multiple Bar Plot
Dataset: Sales data across multiple years/products.

Steps:

Create grouped bars comparing values (e.g., Product A vs Product B across years).

Adjust bar width and alignment.

Add legend, title, and labels.

```python
import matplotlib.pyplot as plt
import numpy as np

years = ["2021", "2022", "2023"]
sales_A = [100, 120, 140]
sales_B = [90, 110, 130]

x = np.arange(len(years))
width = 0.35

plt.bar(x - width/2, sales_A, width, label="Product A")
plt.bar(x + width/2, sales_B, width, label="Product B")

plt.xticks(x, years)
plt.title("Sales Comparison by Year")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.legend()
plt.show()
```
