Task 7: Pie Chart (Market Share)
Dataset: Categorical dataset with proportions (e.g., market share of companies).

Steps:

Create pie chart with category labels.

Display percentage values (autopct).

Add title.

```python
import matplotlib.pyplot as plt

categories = ["A", "B", "C", "D"]
shares = [30, 25, 25, 20]

plt.pie(shares, labels=categories, autopct="%1.1f%%")
plt.title("Market Share")
plt.show()
```
