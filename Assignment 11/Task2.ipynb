###Task 2: Scatter Plot (Relationship)
Dataset: Any dataset with two numerical columns (e.g., height vs weight, marks vs study hours).

Steps:

Create a scatter plot showing relationship between two variables.

Add title and axis labels.

Display the plot.

```python
import matplotlib.pyplot as plt

x = [5, 7, 8, 7, 6, 9, 5, 6, 7, 8]
y = [99, 86, 87, 88, 100, 86, 103, 87, 94, 78]

plt.scatter(x, y)
plt.title("Relationship between Variables")
plt.xlabel("X Variable")
plt.ylabel("Y Variable")
plt.show()
```
