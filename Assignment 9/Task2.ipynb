###Task 2: Mathematical Operations
Goal: Perform element-wise arithmetic.

Arrays:

python
A = np.array([10, 20, 30, 40])
B = np.array([1, 2, 3, 4])
Operations:

Addition, Subtraction, Multiplication, Division

Power (A ** 2)

```python
import numpy as np
A = np.array([10, 20, 30, 40])
B = np.array([1, 2, 3, 4])

print("Task 2 Results:")
print("Addition:", A + B)
print("Subtraction:", A - B)
print("Multiplication:", A * B)
print("Division:", A / B)
print("Power (A**2):", A ** 2)
print()
```
