### Goal: Apply NumPy to a real-world dataset.

Array:

python
sales = np.array([1200, 1500, 900, 2000, 1800, 1700, 1600])
Operations:

Total weekly sales

Average daily sales

Highest & lowest sales day

Standard deviation

Days above average

```python
import numpy as np
sales = np.array([1200, 1500, 900, 2000, 1800, 1700, 1600])

print("Task 7 Results:")
print("Total weekly sales:", np.sum(sales))
print("Average daily sales:", np.mean(sales))
print("Highest sales day:", np.max(sales))
print("Lowest sales day:", np.min(sales))
print("Standard deviation:", np.std(sales))
print("Days above average:", sales[sales > np.mean(sales)])
```
