### Task 4: Aggregation Operations
Goal: Practice aggregation across rows/columns.

Array:

python
data = np.array([
 [10, 20, 30],
 [40, 50, 60],
 [70, 80, 90]
])
Operations:

Row-wise sum

Column-wise sum

Minimum, Maximum

Overall mean

```python
import numpy as np
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Task 4 Results:")
print("Row-wise sum:", np.sum(data, axis=1))
print("Column-wise sum:", np.sum(data, axis=0))
print("Minimum value:", np.min(data))
print("Maximum value:", np.max(data))
print("Overall mean:", np.mean(data))
print()
```
