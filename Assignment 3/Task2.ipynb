Task 2: Recursive Function – Factorial
Function: factorial(n)

Handles:

n == 0 → returns 1

n == 1 → returns 1

Negative → prints error

```python

orders = [1200, 2500, 800, 1750, 3000]

def apply_discounts(order_list):
    """
    Apply discount rules:
    - 10% if order > 2000
    - 5% otherwise
    Prints summary table and returns total revenue.
    """
    print(f"{'Order Amount':<15}{'Discount %':<12}{'Final Amount':<15}")
    total_revenue = 0
    discounted_orders = 0

    for order in order_list:
        discount = 0.1 if order > 2000 else 0.05
        final_amount = order - (order * discount)
        total_revenue += final_amount
        if discount > 0:
            discounted_orders += 1
        print(f"{order:<15}{discount*100:<12}{final_amount:<15.2f}")

    print("\nTotal Revenue after discounts:", total_revenue)
    print("Number of discounted orders:", discounted_orders)
```
