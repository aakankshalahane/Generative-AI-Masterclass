Task 6: Combined Utility Function – Discounts + Filters
Function: process_prices(prices)

Applies 10% discount (map).

Keeps only discounted prices > 500 (filter).

Returns both lists.

```python
# Task 6
def process_prices(prices):
    discounted_prices = list(map(lambda p: p - (p*0.10), prices))
    filtered_prices = list(filter(lambda p: p > 500, discounted_prices))
    return discounted_prices, filtered_prices

print(process_prices([100, 250, 400, 1200, 50, 2000, 850]))
```
