###
Task 2: Recursive Function – Factorial
Function: factorial(n)

Handles:

n == 0 → returns 1

n == 1 → returns 1

Negative → prints error

```python
# Task 2: Recursive Function – Factorial Utility

def factorial(n):
    # Handle negative input
    if n < 0:
        print("Error: Negative number not allowed")
        return None
    
    # Base cases
    if n == 0 or n == 1:
        return 1
    
    # Recursive case
    return n * factorial(n - 1)


# --- Testing the function ---
print("Factorial of 5:", factorial(5))   # Expected 120
print("Factorial of 0:", factorial(0))   # Expected 1
print("Factorial of -3:", factorial(-3)) # Expected error message
```
