Task 7: Mini Problem – Menu Using Functions
Functions:

add_price(prices_list, price)

get_average_price(prices_list)

get_max_price(prices_list)

Menu:

1 → Add price

2 → Show average

3 → Show highest

q → Quit

```python
def add_price(prices_list, price):
    prices_list.append(price)

def get_average_price(prices_list):
    return sum(prices_list)/len(prices_list) if prices_list else 0

def get_max_price(prices_list):
    return max(prices_list) if prices_list else None

print("\nTask 7: Menu System")
prices_list = []
while True:
    print("\nMenu:\n1 - Add price\n2 - Show average\n3 - Show highest\nq - Quit")
    choice = input("Enter choice: ")

    if choice == "1":
        try:
            price = float(input("Enter price: "))
            add_price(prices_list, price)
            print("Price added.")
        except ValueError:
            print("Invalid input.")
            continue
    elif choice == "2":
        print("Average price:", get_average_price(prices_list))
    elif choice == "3":
        print("Highest price:", get_max_price(prices_list))
    elif choice == "q":
        print("Exiting...")
        break
    else:
        print("Invalid choice.")
        continue
```
