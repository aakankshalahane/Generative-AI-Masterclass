Goal: Interactive menu for managing orders.
Logic:
Menu options:
1 → Add order amount to list.
2 → Show all orders + totals (with discounts).
q → Quit program.
Use while True loop:
continue → re-show menu on invalid input.
break → exit on q.
Constraints: Keep logic simple, only lists and loops (no functions).
Output: Interactive session showing orders and totals.
