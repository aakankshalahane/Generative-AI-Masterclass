```python
#Task 4 Loop control with condition
daily = [200, 150, 0, 400, 50, -1, 300]

total_sales = 0

for day in daily:
    if day == -1:
        print("Corrupted data found. Stopping processing.")
        break

    if day == 0:
        continue

    total_sales += day
    print("Running total:", total_sales)

print("Final total:", total_sales)
```
