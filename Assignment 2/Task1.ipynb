### Task 1: Discount Rules (if / elif / else)
Goal: Calculate final order amount after applying discount rules.
Logic:
Input: order_amount (integer).
Discounts:
≥ 2000 → 15%
1500–1999 → 10
1000–1499 → 7%
< 1000 → 0%
Optional: Add 5% tax after discount.
Error Handling: If input is non-numeric, show error and exit.
Output: Subtotal, discount applied, tax (if extra), and final total.

```python
try:
    order_amount = int(input("Enter order amount: "))
except ValueError:
    print("Error: Please enter a valid number.")
    exit()

if order_amount >= 2000:
    discount = 0.15
elif order_amount >= 1500:
    discount = 0.10
elif order_amount >= 1000:
    discount = 0.07
else:
    discount = 0.0

subtotal = order_amount - (order_amount * discount)
tax = subtotal * 0.05
final_total = subtotal + tax

print(f"Order Amount: {order_amount}")
print(f"Discount Applied: {discount*100}%")
print(f"Subtotal after discount: {subtotal}")
print(f"Tax (5%): {tax}")
print(f"Final Total: {final_total}")
```
