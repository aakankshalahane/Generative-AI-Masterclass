```python
# Task 5: Loop Control with Conditions (break & continue)

daily = [200, 150, 0, 400, 50, -1, 300]
total_sales = 0

for sale in daily:
    if sale == -1: 
        print("Corrupted data found. Stopping loop.")
        break
    elif sale == 0:  # no sales → skip
        print("No sales today. Skipping...")
        continue
    else:  # valid positive sales
        total_sales += sale
        print("Running total:", total_sales)

print("Final total sales:", total_sales)
```
