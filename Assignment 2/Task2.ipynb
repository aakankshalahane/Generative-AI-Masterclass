###
Task 2: Process Multiple Orders (for loop)
Goal: Apply Task 1’s discount rules to a list of orders.
Logic:
Orders list: [1200, 2500, 800, 1750, 3000]
Use a for loop to process each order.
Print summary table:
order_amount → discount% → final_amount
Compute total revenue after discounts.
Optional: Count how many orders received a discount.
Output: Table of results + total revenue.

```python
# Initial orders list
orders = [1200, 2500, 800, 1750, 3000]

def apply_discounts(order_list):
    """
    Apply discount rules:
    - 10% if order > 2000
    - 5% otherwise
    Prints summary table and returns total revenue.
    """
    print(f"{'Order Amount':<15}{'Discount %':<12}{'Final Amount':<15}")
    total_revenue = 0
    discounted_orders = 0

    for order in order_list:
        discount = 0.1 if order > 2000 else 0.05
        final_amount = order - (order * discount)
        total_revenue += final_amount
        if discount > 0:
            discounted_orders += 1
        print(f"{order:<15}{discount*100:<12}{final_amount:<15.2f}")

    print("\nTotal Revenue after discounts:", total_revenue)
    print("Number of discounted orders:", discounted_orders)
```
