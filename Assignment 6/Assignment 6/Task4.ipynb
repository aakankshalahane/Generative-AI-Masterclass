###Task 4: File Reader with Exception Handling
Prompts user for a filename and attempts to read it.
Handles FileNotFoundError and PermissionError.
Prints first 3 lines if successful.
finally always prints "File operation attempted".

```python
# Task 4: File Reader with Exception Handling
filename = input("Enter filename: ")

try:
    with open(filename, "r") as f:
        for i in range(3):
            line = f.readline()
            if not line:
                break
            print(line.strip())
except FileNotFoundError:
    print("Error: File not found.")
except PermissionError:
    print("Error: Permission denied.")
finally:
    print("File operation attempted.")
```
