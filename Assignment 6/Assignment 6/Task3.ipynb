###Task 3: Custom Exception — Age Validator
Defines check_age(age) to validate age range (1–120).
Raises ValueError if age is out of range.
Uses try-except to catch and display error messages.
Focus: Custom exception raising and input validation.

```python
# Task 3: Custom Exception: Age Validator
def check_age(age):
    if age < 1 or age > 120:
        raise ValueError("Age must be between 1 and 120")
    return True

try:
    age = int(input("Enter your age: "))
    if check_age(age):
        print("Valid age entered:", age)
except ValueError as e:
    print("Error:", e)
```
