###Task 5: Safe Shopping Cart
Creates an empty cart and repeatedly asks for item prices.
Handles invalid inputs with ValueError and raises error for negatives.
Adds valid prices to cart until user enters 'q'.
Prints total items and total bill at the end.

```python
# Task 5: Safe Shopping Cart
cart = []

while True:
    entry = input("Enter price (or 'q' to quit): ")
    if entry.lower() == 'q':
        break
    try:
        price = float(entry)
        if price < 0:
            raise ValueError("Negative price not allowed")
        cart.append(price)
    except ValueError as e:
        print("Invalid input:", e)

print("Total items:", len(cart))
print("Total bill:", sum(cart))
```
