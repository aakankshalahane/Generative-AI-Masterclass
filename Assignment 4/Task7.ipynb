### Task 7: Export Discounted Prices
- Created a product-price dictionary.
- Asked the user for a discount percentage.
- Calculated discounted prices.
- Saved the results in `discount_report.txt`.
- Calculated total items and average discounted price.


```python
# Task 7: Discounted Prices Report
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

discount = float(input("Enter discount percentage: "))

with open("discount_report.txt", "w") as f:
    for product, price in prices.items():
        discounted = price - (price * discount / 100)
        f.write(f"{product} | {price} | {discounted:.2f}\n")

    discounted_prices = [price - (price * discount / 100) for price in prices.values()]
    f.write(f"\nTotal Items: {len(prices)}\n")
    f.write(f"Average Discounted Price: {sum(discounted_prices)/len(discounted_prices):.2f}\n")

with open("discount_report.txt", "r") as f:
    print("Discount Report:\n", f.read())
```
