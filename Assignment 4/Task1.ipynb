### Task 1: Write Sales Records
- Created a list of sales amounts.
- Wrote sales data into `sales_data.txt`.
- Read and displayed the file contents.

```python
# Task 1: Write Sales Records to a File
sales = [1200, 450, 980, 1500, 3000]

with open("sales_data.txt", "w") as f:
    for s in sales:
        f.write(str(s) + "\n")

# Reopen and print contents
with open("sales_data.txt", "r") as f:
    print("Task 1 Output:\n", f.read())
```
