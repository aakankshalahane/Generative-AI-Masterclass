### Task 3: Append New Sales
- Added new sales to `sales_data.txt`.
- Used append mode (`"a"`).
- Displayed the updated file contents.
- Counted the total number of lines.

```python
# Task 3: Append New Sales
new_sales = [5000, 2500, 1700]

with open("sales_data.txt", "a") as f:
    for s in new_sales:
        f.write(str(s) + "\n")

# Reopen and print updated file
with open("sales_data.txt", "r") as f:
    data = f.readlines()
    print("Updated File:\n", "".join(data))
    print("Total number of lines:", len(data))
```
