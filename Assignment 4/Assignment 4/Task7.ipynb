### Task 7: Export Discounted Prices
- Created a product-price dictionary.
- Asked the user for a discount percentage.
- Calculated discounted prices.
- Saved the results in `discount_report.txt`.
- Calculated total items and average discounted price.


```python
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}
```

```python
# Task 7: Export Discounted Prices

# Ask the user for discount percentage
discount = float(input("Enter discount percentage: "))

discounted_prices = []

# Calculate discounted prices
for product, original_price in prices.items():
    discounted_price = original_price - (original_price * discount / 100)

    discounted_prices.append(discounted_price)

# Write the report into the file
with open("discount_report.txt", "w") as file:
    file.write("Product | Original Price | Discounted Price\n")

    for (product, original_price), discounted_price in zip(
        prices.items(), discounted_prices
    ):
        file.write(
            product + " | " +
            str(original_price) + " | " +
            str(discounted_price) + "\n"
        )

    # Optional summary
    total_items = len(prices)
    average_discounted_price = sum(discounted_prices) / total_items

    file.write("\n")
    file.write("Total Items: " + str(total_items) + "\n")
    file.write("Average Discounted Price: " + str(average_discounted_price))

# Read the file and print it
with open("discount_report.txt", "r") as file:
    report = file.read()

print(report)
```
