### Task 5: Create Product Info File
- Took 3 product names and prices from the user.
- Stored the data in `products.txt`.
- Used the format:
  `ProductName | Price`

```python
# Task 5: Product Info File
products = []
for i in range(3):
    name = input(f"Enter product {i+1} name: ")
    price = input(f"Enter price of {name}: ")
    products.append(f"{name} | {price}")

with open("products.txt", "w") as f:
    for p in products:
        f.write(p + "\n")

with open("products.txt", "r") as f:
    print("Products File Content:")
    for line in f:
        print(line.strip())
```
